# 14. Какая команда выиграла наибольшое количество матчей в сезоне? (Будьте внимательны с типом матча в сезоне - season).

```python
regular = df[df['season'] == 'regular season']

home_wins = regular[regular['home_team_win'] == 1]['home_team'].value_counts()
away_wins = regular[regular['home_team_loss'] == 1]['away_team'].value_counts()

(home_wins + away_wins).sort_values(ascending=False)
```

Тут по заданию говорится про колонку season, в данных есть regular season и post season. Считаю только регулярный, поэтому сначала фильтровал его в переменную regular.Потом победы дома (home_team), где home_team_win = 1, а победы в гостях - это away_team там, где домашняя команда проиграла (home_team_loss = 1). Складываем обе серии и сортируем по убыванию через sort_values(ascending=False), чтобы лидер был сверху.

Ковариация получилась положительная (около 225) , значит зависимость есть: чем больше зрителей, тем чаще домашняя команда выигрывает. Значение небольшое, так что связь слабая, но она есть.
